```python
import re
from typing import List, Dict
import nltk
from sentence_transformers import SentenceTransformer
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans

nltk.download('punkt', quiet=True)

# Load embedding model once
EMBED_MODEL_NAME = "all-MiniLM-L6-v2"
embed_model = SentenceTransformer(EMBED_MODEL_NAME)

# ---------- Helper tokenizers ----------
def word_tokenize(text: str) -> List[str]:
    return nltk.tokenize.word_tokenize(text)

def sentence_tokenize(text: str) -> List[str]:
    return nltk.tokenize.sent_tokenize(text)

def paragraph_split(text: str) -> List[str]:
    parts = re.split(r'\n\s*\n', text.strip())
    parts = [p.strip() for p in parts if p.strip()]
    return parts

# ---------- 1. Sliding window (words) ----------
def chunk_sliding_window_by_words(text: str, chunk_words: int = 120, overlap_words: int = 20) -> List[Dict]:
    words = word_tokenize(text)
    step = chunk_words - overlap_words
    if step <= 0:
        raise ValueError("chunk_words must be > overlap_words")
    chunks = []
    i = 0
    cid = 0
    while i < len(words):
        slice_words = words[i : i + chunk_words]
        chunks.append({"id": cid, "text": " ".join(slice_words), "start_word": i, "end_word": min(i+chunk_words, len(words))})
        cid += 1
        i += step
    return chunks

# ---------- 2. Sentence accumulation (coherent chunks) ----------
def chunk_by_sentence_accumulation(text: str, max_words: int = 120, overlap_words: int = 20) -> List[Dict]:
    sentences = sentence_tokenize(text)
    chunks = []
    current = []
    current_words = 0
    cid = 0
    i = 0
    while i < len(sentences):
        s = sentences[i]
        s_words = len(word_tokenize(s))
        # if single sentence bigger than max_words and empty current, accept it solo
        if s_words > max_words and not current:
            chunks.append({"id": cid, "text": s, "start_sent": i, "end_sent": i})
            cid += 1
            i += 1
            continue
        if current_words + s_words > max_words and current:
            chunks.append({"id": cid, "text": " ".join(current)})
            cid += 1
            # prepare overlap
            if overlap_words > 0:
                last_words = word_tokenize(" ".join(current))[-overlap_words:]
                current = [" ".join(last_words)] if last_words else []
                current_words = len(last_words)
            else:
                current = []
                current_words = 0
        else:
            current.append(s)
            current_words += s_words
            i += 1
    if current:
        chunks.append({"id": cid, "text": " ".join(current)})
    return chunks

# ---------- 3. Paragraph chunking ----------
def chunk_by_paragraphs(text: str, max_words: int = None) -> List[Dict]:
    paras = paragraph_split(text)
    chunks = []
    cid = 0
    for p in paras:
        words = word_tokenize(p)
        if max_words is None or len(words) <= max_words:
            chunks.append({"id": cid, "text": p})
            cid += 1
        else:
            # split long paragraph by sliding window
            sub = chunk_sliding_window_by_words(p, chunk_words=max_words, overlap_words=int(max_words*0.1))
            for s in sub:
                s['id'] = cid
                chunks.append(s)
                cid += 1
    return chunks

# ---------- 4. Semantic clustering chunking ----------
def chunk_semantic_clustering(text: str, n_clusters: int = 8) -> List[Dict]:
    sentences = sentence_tokenize(text)
    if not sentences:
        return []
    sent_embs = embed_model.encode(sentences, convert_to_numpy=True, show_progress_bar=False)
    k = min(n_clusters, len(sentences))
    km = KMeans(n_clusters=k, random_state=42, n_init='auto')
    labels = km.fit_predict(sent_embs)
    # group sentences by label while preserving original order
    buckets = {}
    idxs = {}
    for i, lab in enumerate(labels):
        buckets.setdefault(lab, []).append(sentences[i])
        idxs.setdefault(lab, []).append(i)
    # order clusters by earliest sentence in that cluster
    cluster_order = sorted([(lab, min(idxs[lab])) for lab in idxs], key=lambda x: x[1])
    chunks = []
    cid = 0
    for lab, _ in cluster_order:
        chunks.append({"id": cid, "text": " ".join(buckets[lab]), "cluster": int(lab), "sent_count": len(buckets[lab])})
        cid += 1
    return chunks

# ---------- Embedding + retrieval helpers ----------
def embed_texts(texts: List[str]) -> np.ndarray:
    if not texts:
        return np.zeros((0, embed_model.get_sentence_embedding_dimension()))
    return embed_model.encode(texts, convert_to_numpy=True, show_progress_bar=False)

def retrieve_top_k(chunks: List[Dict], query: str, k: int = 3) -> List[Dict]:
    texts = [c['text'] for c in chunks]
    emb_chunks = embed_texts(texts)
    q_emb = embed_texts([query])[0].reshape(1, -1)
    scores = cosine_similarity(emb_chunks, q_emb).reshape(-1)
    top_idx = np.argsort(scores)[::-1][:k]
    results = []
    for i in top_idx:
        c = chunks[i].copy()
        c['score'] = float(scores[i])
        results.append(c)
    return results
```

```python
"""
Token-based chunking + Azure OpenAI embeddings + retrieval
Uses your exact AzureOpenAI SDK style.
"""

import os
import numpy as np
from dotenv import load_dotenv
from typing import List, Dict
from sklearn.metrics.pairwise import cosine_similarity

# ---------------------------------------
# Load Azure OpenAI Client
# ---------------------------------------
from openai import AzureOpenAI

load_dotenv(override=True)

client = AzureOpenAI(
    api_key=os.environ["AZURE_OPENAI_KEY"],
    azure_endpoint=os.environ["AZURE_OPENAI_ENDPOINT"],
    api_version="2023-10-01-preview"
)

# Embedding deployment name
DEPLOYMENT = os.environ["OPENAI_DEPLOYMENT_NAME"]


# ---------------------------------------
# Tokenizer (tiktoken)
# ---------------------------------------
try:
    import tiktoken
    ENC = tiktoken.get_encoding("cl100k_base")
except Exception:
    ENC = None
    print("WARNING: tiktoken not installed. Install: pip install tiktoken")


def count_tokens(text: str) -> int:
    """Return token count using tiktoken; fallback to words."""
    if ENC:
        return len(ENC.encode(text))
    return len(text.split())


# ---------------------------------------
# Chunking (Token-based)
# ---------------------------------------
def chunk_by_tokens(text: str, max_tokens: int = 300, overlap: int = 50) -> List[Dict]:
    """
    Token-aware chunking used in RAG pipelines.
    - Uses tiktoken for exact LLM token boundaries.
    """
    if ENC is None:
        # fallback: word splitting
        words = text.split()
        step = max_tokens - overlap
        chunks = []
        i = 0
        cid = 0
        while i < len(words):
            sub = words[i:i+max_tokens]
            chunks.append({
                "id": cid,
                "text": " ".join(sub)
            })
            cid += 1
            i += step
        return chunks

    # Real token-aware chunking
    ids = ENC.encode(text)
    n = len(ids)
    step = max_tokens - overlap

    chunks = []
    cid = 0
    i = 0

    while i < n:
        window_ids = ids[i:i+max_tokens]
        chunk_text = ENC.decode(window_ids)
        print(f"chunk_text after decode=====: {chunk_text}")
        chunks.append({
            "id": cid,
            "text": chunk_text,
            "token_start": i,
            "token_end": i + len(window_ids)
        })
        cid += 1
        i += step

    return chunks


# ---------------------------------------
# Azure OpenAI Embeddings
# (Your exact SDK style)
# ---------------------------------------
def get_embeddings(texts: List[str]) -> np.ndarray:
    """
    Returns np.ndarray of shape (N, dim)
    """
    resp = client.embeddings.create(
        model=DEPLOYMENT,
        input=texts
    )
    embs = [np.array(item.embedding, dtype=np.float32) for item in resp.data]
    return np.vstack(embs)


# ---------------------------------------
# Build vector index
# ---------------------------------------
def build_index(emb_matrix: np.ndarray) -> np.ndarray:
    """
    Normalize embeddings for cosine similarity.
    """
    norms = np.linalg.norm(emb_matrix, axis=1, keepdims=True)
    norms[norms == 0] = 1.0
    return emb_matrix / norms


# ---------------------------------------
# Retrieval
# ---------------------------------------
def retrieve_top_k(index: np.ndarray, chunks: List[Dict], query: str, k: int = 5):
    q_emb = get_embeddings([query])
    q_emb = q_emb / (np.linalg.norm(q_emb, axis=1, keepdims=True) + 1e-12)

    sims = (index @ q_emb.T).reshape(-1)
    top_idx = np.argsort(sims)[::-1][:k]

    results = []
    for idx in top_idx:
        results.append({
            "chunk_id": chunks[idx]["id"],
            "score": float(sims[idx]),
            "text": chunks[idx]["text"][:300]  # preview
        })
    return results


# ---------------------------------------
# Demo Run
# ---------------------------------------
if __name__ == "__main__":

    # Load long text corpus
    with open("sample_long_text.txt", "r", encoding="utf-8") as f:
        text = f.read()

    print("Chunking...")
    chunks = chunk_by_tokens(text, max_tokens=300, overlap=60)
    chunk_texts = [c["text"] for c in chunks]
    print(f"Created {len(chunks)} chunks.")

    print("\nEmbedding chunks via Azure OpenAI...")
    emb = get_embeddings(chunk_texts)
    index = build_index(emb)

    # Try a retrieval query
    query = "How does transformer architecture improve NLP?"
    print("\nRetrieving top passages:\n")

    results = retrieve_top_k(index, chunks, query, k=3)
    for r in results:
        print(f"Chunk {r['chunk_id']}, Score={r['score']:.4f}")
        print(r["text"])
        print("-" * 80)
```
